```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df[features])
```

```python
# Elbow method to choose k
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(x_scaled)
    wcss.append(km.inertia_)
```

```python
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.title('Elbow method')
plt.show()
```

```python
km_model = KMeans(n_clusters=6, random_state=42, n_init=10)
km_model.fit(x_scaled)
df['cluster_group'] = km_model.labels_
df['cluster_group'].value_counts()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18,7))
sns.scatterplot(x='energy', y='danceability', hue='cluster_group', data=df, palette='tab10', s=8, ax=axes[0])
axes[0].set_title('Clusters: energy vs danceability')
sns.scatterplot(x='acousticness', y='speechiness', hue='cluster_group', data=df, palette='tab10', s=8, ax=axes[1])
axes[1].set_title('Clusters: acousticness vs speechiness')
plt.show()
```

```python
# What each cluster sounds like (average of each feature)
plt.figure(figsize=(14,5))
sns.heatmap(df.groupby('cluster_group')[features].mean().round(2), annot=True, cmap='YlGnBu', fmt='g')
plt.title('Average features in each cluster')
plt.show()
```

```python
# Clusters vs playlist genre
genre_cluster = pd.crosstab(df['playlist_genre'], df['cluster_group'], normalize='index')
sns.heatmap(genre_cluster, annot=True, fmt='.0%', cmap='Greens')
plt.title('Playlist genre vs cluster')
plt.show()
```

```python
# Clusters vs playlist name (top 15 playlists)
top_pl = df['playlist_name'].value_counts().head(15).index
pl_cluster = pd.crosstab(df[df['playlist_name'].isin(top_pl)]['playlist_name'], df['cluster_group'], normalize='index')
plt.figure(figsize=(10,8))
sns.heatmap(pl_cluster, annot=True, fmt='.0%', cmap='Purples')
plt.title('Top 15 playlists vs cluster')
plt.show()
```

```python
x = df[['danceability','energy','key','loudness','mode','speechiness','acousticness',
        'instrumentalness','liveness','valence','tempo','duration_ms','track_popularity','release_year']]
y = df['playlist_genre']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print(len(x_train))
print(len(x_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc_model = RandomForestClassifier(n_estimators=300, random_state=42)
rfc_model.fit(x_train, y_train)
```

```python
ans = rfc_model.predict(x_test)    # predicted genres
ans
```

```python
accuracy = rfc_model.score(x_test, y_test)
```

```python
print('Accuracy :', round(accuracy * 100, 2), '%')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, ans))
```

```python
from sklearn.metrics import confusion_matrix
labels = rfc_model.classes_
cm = confusion_matrix(y_test, ans, labels=labels)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=labels, yticklabels=labels, cmap='Greens')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion matrix – Random Forest')
plt.show()
```

```python
importance = pd.Series(rfc_model.feature_importances_, index=x.columns).sort_values()
importance.plot(kind='barh', color='green', figsize=(8,6))
plt.title('Which features matter most for the genre?')
plt.show()
```

```python
genre_prob = rfc_model.predict_proba(x)              # probability of each genre for every song
```

```python
song_vectors = np.hstack([x_scaled, genre_prob * 3]) # *3 gives the genre fingerprint more weight
```

```python
def recommend(song_name, n=10):
    match = df[df['track_name'].str.lower() == song_name.lower()]
    if len(match) == 0:
        print('Song not found. Use search_song() to find the exact name.')
        return
    i = match['track_popularity'].idxmax()
    print('Because you like:', df.loc[i, 'track_name'], '-', df.loc[i, 'track_artist'],
          '| genre:', df.loc[i, 'playlist_genre'], '| cluster:', df.loc[i, 'cluster_group'])

    same_cluster = df[df['cluster_group'] == df.loc[i, 'cluster_group']].index
    dist = np.linalg.norm(song_vectors[same_cluster] - song_vectors[i], axis=1)   # euclidean distance
    result = df.loc[same_cluster, ['track_name','track_artist','playlist_genre','track_popularity']].copy()
    result['distance'] = dist
    base = df.loc[i, 'track_name'].lower().split(' - ')[0]
    result = result[result['track_name'].str.lower().str.split(' - ').str[0] != base]   # remove the same song / remixes
    return result.sort_values('distance').head(n).reset_index(drop=True)
```

```python
def search_song(text):
    return df[df['track_name'].str.contains(text, case=False)][['track_name','track_artist','playlist_genre']].head(10)

recommend('Memories')

search_song('believer')

recommend('Believer')
```
